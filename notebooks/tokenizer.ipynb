# Privan-130M: Byte-Level BPE Tokenizer Walkthrough

This notebook demonstrates:
1. Training a Byte-Level Byte-Pair Encoding (BPE) Tokenizer from scratch
2. Managing special tokens (`<PAD>`, `<UNK>`, `<BOS>`, `<EOS>`)
3. Encoding strings into discrete token IDs and decoding back to text
4. Handling arbitrary UTF-8 bytes without out-of-vocabulary exceptions

```python
import sys
from pathlib import Path

# Add src to Python path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from llm.tokenizer import ByteLevelBPETokenizer
```

## 1. Load or Train the Tokenizer

```python
tokenizer_path = Path("../data/tokenizer")
if tokenizer_path.exists():
    tokenizer = ByteLevelBPETokenizer.load(tokenizer_path)
    print(f"Loaded existing tokenizer with vocab size: {tokenizer.vocab_size():,}")
else:
    print("Tokenizer not yet trained. Run scripts/train_tokenizer.py first.")
```

## 2. Inspect Special Tokens

```python
print("PAD Token ID:", tokenizer.pad_token_id)
print("UNK Token ID:", tokenizer.unk_token_id)
print("BOS Token ID:", tokenizer.bos_token_id)
print("EOS Token ID:", tokenizer.eos_token_id)
```

## 3. Encode and Decode Samples

```python
sample_text = "Artificial intelligence and Transformers revolutionize language modeling."
token_ids = tokenizer.encode(sample_text)
decoded_text = tokenizer.decode(token_ids)

print("Original: ", sample_text)
print("Token IDs:", token_ids)
print("Decoded:  ", decoded_text)
assert sample_text == decoded_text, "Roundtrip mismatch!"
```
